# 🎯 YOLOv8 Custom Object Detection

## 📌 Project Overview

YOLOv8 (You Only Look Once Version 8) is one of the fastest and most accurate real-time object detection algorithms developed by Ultralytics. It can detect multiple objects in a single image while maintaining high speed and accuracy.

In this project, a custom YOLOv8 model is trained to classify two classes:

- Bald
- No Bald

The dataset consists of annotated images divided into training, validation, and testing sets. Each image has a corresponding label file in YOLO annotation format.

After training, the model will be evaluated and used to detect objects on unseen images.

# 📚 Import Required Libraries

The following libraries are required for loading the YOLO model, training, prediction, image processing, and visualization.

```python
from ultralytics import YOLO
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the Pre-trained YOLOv8 Model
model = YOLO("yolov8n.pt")
```

# 📂 Understanding the Dataset Configuration

The dataset is configured using a YAML file that specifies:

- Training image directory
- Validation image directory
- Testing image directory
- Number of object classes
- Class names

YOLO automatically reads this configuration during training.

```python
import yaml

dataset_yaml = "dataset.yaml"

with open(dataset_yaml, "r") as file:
    config = yaml.safe_load(file)

config
```

# 🏋️ Train the YOLOv8 Model

The YOLOv8 model is trained using the custom dataset defined in the YAML configuration file.

During training, the model learns to identify object locations (bounding boxes) and classify them into the specified object classes.

The training process uses transfer learning, where a pre-trained YOLOv8 model is fine-tuned on the custom dataset.

```python
results = model.train(
    data="dataset.yaml",
    epochs=50,
    imgsz=640,
    batch=8,
    project="output",
    name="Bald_Detection"
)
```

# 📊 Validate the Trained Model

After training, the model is evaluated on the validation dataset to measure its performance. During validation, YOLO computes important metrics such as Precision, Recall, mAP@50, and mAP@50-95.

```python
metrics = model.val()
```

# 🖼️ Predict on a Test Image

The trained model is now used to detect objects in a new image. YOLO predicts object locations, class names, and confidence scores.

```python
results = model.predict(
    source="Data/test_Dataset/images/test/sithuimages14.jpeg",
    conf=0.25,
    save=True
)
```

# 🏆 Load the Best Trained Model

After training, YOLO automatically saves the model with the best validation performance as **best.pt**. This trained model is used for making predictions on unseen images.

```python
best_model = YOLO("runs/detect/output/Bald_Detection-3/weights/best.pt")
```

# 🖥️ Display the Prediction

The predicted image with bounding boxes is displayed for visualization.

```python
results[0].show()
```

# 🖼️ Perform Object Detection on a Test Image

The trained YOLOv8 model is used to detect objects in an unseen test image. The model predicts the object class, confidence score, and bounding box location.

```python
results = best_model.predict(
    source="Data/test_Dataset/images/test/tonyimages24.jpeg",
    conf=0.25,
    save=True
)
```

# 📈 Training Results

During training, YOLOv8 automatically generated several evaluation graphs:

- **results.png** – Shows the training and validation losses over epochs.
- **confusion_matrix.png** – Displays the model's classification performance.
- **F1_curve.png** – Shows the relationship between confidence threshold and F1 score.
- **P_curve.png** – Precision at different confidence thresholds.
- **R_curve.png** – Recall at different confidence thresholds.
- **PR_curve.png** – Precision-Recall curve used to evaluate object detection performance.

These graphs help assess how well the model learned and whether it generalizes effectively to unseen data.

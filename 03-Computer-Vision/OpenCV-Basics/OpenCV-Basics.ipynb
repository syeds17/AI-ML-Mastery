```python
#Importing Lib
import cv2
import os
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Loading Data set
image = cv2.imread("assets/baby.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
#Assigining Output Folder
output_folder = "outputs"
os.makedirs(output_folder, exist_ok=True)
```

```python
#Performing Operations on Image
h_flip = cv2.flip(image, 1)                                       #  1. Horizontal Flip

v_flip = cv2.flip(image, 0)                                       # 2. Vertical Flip

rot90 = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)                # 3. Rotation (90 degrees )

blur = cv2.GaussianBlur(image, (7, 7), 0)                        # 4. Gaussian Blur

bright = cv2.convertScaleAbs(image, alpha=1.2, beta=30)          # 5. Brightness Adjustment
```

```python
# Show Original + Augmentations
titles = ["Original", "Horizontal Flip", "Vertical Flip", "Rotate 90°", "Gaussian Blur",  "Brightness ↑"]
images = [image, h_flip, v_flip, rot90, blur, bright]
```

```python
#Plotting the output
plt.figure(figsize=(12, 8))

for i, img in enumerate(images):
    plt.subplot(2, 3, i + 1)
    plt.imshow(img)
    plt.title(titles[i])
    plt.axis("off")

    save_path = os.path.join(output_folder, f"{titles[i].replace(' ', '_')}.jpg")
    cv2.imwrite(save_path, cv2.cvtColor(img, cv2.COLOR_RGB2BGR))   # convert back to BGR for saving

plt.tight_layout()
plt.show()

print(f" Augmented images saved in folder: {output_folder}")
```

```python
#Import lib
import cv2
```

```python
#Load the Sample Video
video_path = "assets/sample_video.mp4"
cap = cv2.VideoCapture(video_path)
```

```python
#Verify Video Loading
if cap.isOpened():
    print("Video loaded successfully!")
else:
    print("Error loading video!")
```

```python
#Read Video Frame by Frame
while True:
    ret, frame = cap.read()

    if not ret:
        break

    cv2.imshow("Video", frame)

    if cv2.waitKey(25) & 0xFF == ord('q'):
        break
```

```python
#Release Video Resources
cap.release()
cv2.destroyAllWindows()
```

# 📷 Capture Live Video from Webcam

Instead of reading frames from a video file, OpenCV can capture frames directly from a webcam. Each frame is processed and displayed in real time until the user exits the application.

```python
cap = cv2.VideoCapture(0)
frame_count = 0
max_frames = 300 

if cap.isOpened():
    print("Webcam opened successfully!")
else:
    print("Failed to open webcam.")
```

```python
#Display Live Webcam Feed
while frame_count < max_frames:
    ret, frame = cap.read()

    if not ret:
        print("Failed to capture frame.")
        break

    cv2.imshow("Live Webcam", frame)
    frame_count += 1

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
```

```python
#Release Webcam Resources
cap.release()
cv2.destroyAllWindows()
```

# 💾 Record and Save Webcam Video

OpenCV provides the `VideoWriter` class to save frames captured from a webcam or processed video into a video file. This allows users to record live webcam footage and store it for future use.

```python
#Import Lib
import cv2
```

```python
#Accessing WebCam
cap = cv2.VideoCapture(0)
fourcc = cv2.VideoWriter_fourcc(*'XVID')
```

```python
#Output Video
out = cv2.VideoWriter(
    "outputs/recorded_video.avi",
    fourcc,
    20.0,
    (640,480)
)
```

```python
#Recording Video
frame_count = 0
max_frames = 300      

while frame_count < max_frames:
    ret, frame = cap.read()

    if not ret:
        break

    out.write(frame)

    cv2.imshow("Recording...", frame)

    frame_count += 1

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
```

```python
#Release WebCam
cap.release()
out.release()
cv2.destroyAllWindows()
print("Video saved successfully!")
```

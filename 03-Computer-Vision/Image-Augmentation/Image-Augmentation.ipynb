# 🖼️ Image Augmentation using TensorFlow/Keras

## 📌 Project Overview

Image Augmentation is a Computer Vision technique used to artificially increase the size and diversity of an image dataset by applying different transformations.

Instead of collecting thousands of new images, we can generate new variations of existing images such as rotated, flipped, zoomed, shifted, and brightness-adjusted images.

Image augmentation helps improve the performance and generalization of Deep Learning models while reducing overfitting.

---

## 🎯 Objective

- Understand Image Augmentation.
- Learn different augmentation techniques.
- Generate multiple versions of the same image.
- Improve dataset diversity for Deep Learning.

# 📚 Import Required Libraries

The following libraries are required for loading images, performing augmentation, and visualizing the results.

```python
#Importing Lib
from tensorflow.keras.preprocessing.image import ImageDataGenerator, img_to_array, load_img
import matplotlib.pyplot as plt
import numpy as np
import os
```

```python
#Loading and converting Image into array
img = load_img("Data/dog.jpg")
image_arr = img_to_array(img)
image_arr = np.expand_dims(image_arr, axis=0)
```

```python
#Initalizing output folder
output_folder = "outputs"
os.makedirs(output_folder, exist_ok=True)
```

```python
#Functions Applied on the Image
datagen = ImageDataGenerator(
    horizontal_flip = True,
    vertical_flip = True,
    rotation_range = 90,
    brightness_range = [0.5,1.5],
    zoom_range = 0.1
)
```

```python
#Performing Augmentation
aug_iter = datagen.flow(
    image_arr,
    batch_size = 1,
    save_to_dir = output_folder,
    save_prefix = "aug",
    save_format = "jpg"
)
```

```python
#Original Image
plt.figure(figsize=(12, 8))
plt.subplot(2, 3, 1)
plt.imshow(img)
plt.title("Original")
plt.axis("off")
```

```python
for i in range(5):
    batch = next(aug_iter)                                  
    aug_img = batch[0].astype("uint8")
    plt.subplot(2, 3, i + 2)
    plt.imshow(aug_img)
    plt.title(f"Augmented {i+1}")
    plt.axis("off")
plt.tight_layout()
plt.show()

print(f" Augmented images saved in: {output_folder}")
```

# 📉 Gradient Descent Algorithm from Scratch

## 📌 Project Overview

Gradient Descent is one of the most fundamental optimization algorithms in Machine Learning and Deep Learning. It is used to minimize a cost (loss) function by iteratively updating the model parameters in the direction of the steepest decrease.

In this project, Gradient Descent is implemented **from scratch using Python** without relying on any machine learning libraries. A simple mathematical function is optimized to demonstrate how the algorithm gradually approaches the minimum value.

---

## 🎯 Objective

- Understand the intuition behind Gradient Descent.
- Learn how gradients are calculated.
- Implement Gradient Descent from scratch.
- Visualize the optimization process.
- Observe how different learning rates affect convergence.

```python
import matplotlib.pyplot as plt

#Initialize the parameters
theta = 5.0                   #Initial value of theta
learning_rate = 0.1           #Step size for each iteration 
iterations = 50

#Gradient Function
def grad(theta):
    return 2 * theta  # Derivative of the cost function (theta^2)

#Cost Function
def cost(theta):
    return theta ** 2  # Cost function (theta^2)

theta_history = []  # To store the history of theta values
cost_history = []   # To store the history of cost values
iteration_history = []  # To store the history of iterations

# Gradient Descent Algorithm
for i in range(iterations):
    theta_history.append(theta)  
    cost_history.append(cost(theta))  
    iteration_history.append(i + 1)  # Store the current iteration number
    theta -= learning_rate * grad(theta) # Update theta using the Gradient Descent update rule

#Plotting the results
plt.figure(figsize=(10, 5))
plt.plot(iteration_history, cost_history, marker='o',linewidth=2, color='b')
plt.title('Gradient Descent Convergence for J(θ) = θ²')   
plt.xlabel('Iterations')
plt.ylabel('Cost J(θ)')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()  
```

```python
#Creating an iteration Table to visualize the results
import pandas as pd

history = pd.DataFrame({
    "Iteration": range(1, iterations + 1),
    "Theta": theta_history,
    "Cost": cost_history
})

history.head()
```

```python
# Printing the final results
print(f"Final Theta : {theta:.6f}")
print(f"Final Cost  : {cost(theta):.6f}")
```

```python
# Plotting Theta Convergence
plt.figure(figsize=(10,5))
plt.plot(theta_history)
plt.title("Theta Convergence")
plt.xlabel("Iterations")
plt.ylabel("Theta")
plt.grid(alpha=0.3)
plt.show()
```
